```python
from sklearn.linear_model import LogisticRegression  #引入逻辑回归器
```

```python
model=LogisticRegression(
    solver='sag',           #选择优化算法：sag 为随机平均梯度法，适合样本量较大的场景
    max_iter=1000,
    class_weight='balanced',     #用于设置各类别的损失权重，balanced 根据样本量自动加权
    random_state=42,
    C=1.0,       #C 是正则化强度的倒数，C 越小正则化越强；sklearn 1.8+ 用 C 控制 L2，无需再写 penalty
    #注意：sklearn 1.7 起已删除 multi_class 参数，solver='sag' 的多分类默认就是 multinomial 策略
)                  #实例化一个逻辑回归模型
```

```python
#OVR  多分类策略，选择一个标签为正例，其他标签为反例，有k组分类；取得置信分最高的类别作为当前的二分类
#1.直接创建logisticregression对象（已经弃用）
# model_ovr1=LogisticRegression(multi_class='ovr')

#2.创建OneVsRestClassifier模型
from sklearn.multiclass import OneVsRestClassifier   #从多分类模块引入ovr多分类器
model_ovr2=OneVsRestClassifier(estimator=LogisticRegression())   #需要指定基础分类器，也就是真正干活的二分类模型
```
