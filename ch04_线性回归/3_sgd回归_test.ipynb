```python
import sklearn
sklearn.set_config(display="text")  # Windows 中文环境下修复: 避免 estimator 的 HTML 摘要在读 estimator.js 时报 UnicodeDecodeError
from sklearn.linear_model import SGDRegressor  #随机梯度下降，每次只取一个样本计算梯度，也就是只取一个样本计算损失函数再求梯度，批量梯度下降是算整体样本损失函数的求和再求梯度
```

```python
#1.定义数据
#自变量，每周学习时间
x=[[5],[8],[10],[12],[15],[3],[7],[9],[14],[6]]
#因变量，数学考试成绩
y=[55,65,70,75,85,50,60,72,80,58]
```

```python
#2.创建随机梯度度下降回归模型（拟合一个函数,让损失函数最小）
sgd=SGDRegressor(
    penalty='l2',
    loss='squared_error',   #损失函数：默认是均方误差
    max_iter=10000000,   #最大迭代次数，默认是1000
    eta0=0.00001,     #学习率，也就是步长alpha
    learning_rate='constant',  #常数学习率，设置学习率不变
    tol=1e-8   #收敛阈值，也就是相邻两次损失函数的差的绝对值小于该值时停止
)  #实例化sgd回归对象

#3.模型训练
sgd.fit(x,y)

print(sgd.coef_)
print(sgd.intercept_)
```
