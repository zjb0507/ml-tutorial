```python
import sklearn
sklearn.set_config(display='text')  # 关闭 HTML 图示渲染，规避中文 Windows(GBK) 下的 UnicodeDecodeError

from sklearn.linear_model import LinearRegression
```

```python
#1.定义数据
#自变量，每周学习时间
x=[[5],[8],[10],[12],[15],[3],[7],[9],[14],[6]]
#因变量，数学考试成绩
y=[55,65,70,75,85,50,60,72,80,58]
```

```python
#2.创建模型：线性回归
lrmodel=LinearRegression()  #实例化线性回归模型对象
```

```python
#3.模型训练
lrmodel.fit(x,y)
```

```python
#4.打印模型参数
print(lrmodel.coef_)  #斜率
print(lrmodel.intercept_)  #截距
```

```python
#5.预测
x_new=[[11],[12],[10.4]]
y_pred=lrmodel.predict(x_new)
print(y_pred)
```

```python
#6.画图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = "SimHei"
rcParams['axes.unicode_minus'] = False   # 负号用 ASCII 减号，避免 SimHei 缺字显示成方块

#注意：list 的 extend() 是"就地修改"，返回值是 None，不能用 x_list = x.extend(...)
x.extend(x_new)
y.extend(y_pred)
plt.scatter(x, y, label='时间—成绩', c='r')
plt.xlabel('每周学习时间')
plt.ylabel('数学考试成绩')
plt.legend(loc='upper left')
plt.title('学习时间与分数关系散点图')
plt.grid(True)
plt.show()
```

```python
#用方差和协方差验证数学求解公式（计算线性相关函数的系数，也就是最小二乘法）
#斜率等于x和y的协方差除以x的方差，截距等于y均值-斜率*x的均值
import numpy as np
x=np.array(x).reshape(-1)
cov=np.cov(x,y)  #计算x，y的方差及协方差
print(cov)
beta1=cov[0,1]/cov[0,0]  #计算斜率
beta2=np.mean(y)-beta1*np.mean(x)    #计算截距
print(beta1,beta2)
```
