```python
#奇异值分解，用来特征降维
import numpy as np
```

```python
A=np.array([[1,1],[2,2],[0,0]])

U,S,V=np.linalg.svd(A)    #调用线性代数模块里的奇异值分解函数

print(U)
print(S)
print(V)
```

```python
from sklearn.utils.extmath import randomized_svd     #调用sklearn中的随机截断svd
U,S,V=randomized_svd(A,n_components=2)   #指定提取几列主成分特征
print(U)
print(S)
print(V)
```

```python
from sklearn.decomposition import PCA

x=np.random.randn(100,3)
pca=PCA(n_components=2)
x_pca=pca.fit_transform(x)   #对原始特征进行主成分分析，降维
print(x_pca.shape)
```
